```python
import numpy as np
import pandas as pd
```

```python
#Loading data
df1=pd.read_csv("Day9_Customers.csv")
print(df1)
```

```python
df2=pd.read_csv("Day9_Orders.csv")
print(df2)
```

```python
df3=pd.read_csv("Day9_Products.csv")
print(df3)
```

```python
#combining related information
#using merge()
#we can only merge 2 tables if they have common column between them
df4=pd.merge(df1,df2,on="Customer_ID")
print(df4)
```

```python
#combining all 3 csv files
df5=pd.merge(df4,df3,on="Product_ID")
print(df5)
```

```python
#LEFT MERGE
df6=pd.merge(df1,df2,on="Customer_ID",how="left")
print(df6)
```

```python
#inner merge
df7=pd.merge(df1,df2,on="Customer_ID",how="inner")
print(df7)
```

```python
#right merge
df8=pd.merge(df1,df2,on="Customer_ID",how="right")
print(df8)
```

```python
#outer merge
df9=pd.merge(df1,df2,on="Customer_ID",how="outer")
print(df9)
```

```python
#using concat()
combined=pd.concat([df1,df2,df3],ignore_index=True,axis=1)
print(combined)
```

```python
#apply() and lambda
def total_sales(row):
    return row["Quantity"] * row["Unit_Price"]

df5["Total_Sales"] = df5.apply(total_sales, axis=1)
df5
```

```python
#converting to datetime
df5["Order_Date"] = pd.to_datetime(df5["Order_Date"])
df5
```

```python
df5["Order_Year"] = df5["Order_Date"].dt.year
df5
```

```python
df5["Order_Month"] = df5["Order_Date"].dt.month
df5
```

```python
df5["Order_Day"] = df5["Order_Date"].dt.day
df5
```

```python
df5["Order_Day_Name"] = df5["Order_Date"].dt.day_name()
df5
```

```python
df12 = df5[
    [
        "Order_ID",
        "Order_Date",
        "Order_Year",
        "Order_Month",
        "Order_Day",
        "Order_Day_Name",
        "Customer_ID",
        "Customer_Name",
        "City",
        "Region",
        "Membership_Type",
        "Product_ID",
        "Product_Name",
        "Category",
        "Brand",
        "Unit_Price",
        "Quantity",
        "Total_Sales",
        "Payment_Method",
        "Order_Status"
    ]
]

df12
```

```python
df12.to_csv("Day9_Processed_Orders.csv", index=False)
```
